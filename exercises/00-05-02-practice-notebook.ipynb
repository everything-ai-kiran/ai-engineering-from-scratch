# Load CSV and show dataframe

```python
import pandas as pd

df = pd.read_csv('data.csv')
df
```

```python
import matplotlib.pyplot as plt

# Plot salary trend by employee ID
df_plot = df.copy()
df_plot['employee_num'] = df_plot['employee_id'].str.replace('E', '', regex=False).astype(int)
# df_plot = df_plot.sort_values('employee_num')

plt.figure(figsize=(12, 5))
plt.plot(df_plot['employee_num'], df_plot['salary'], marker='o', linewidth=1)
plt.title('Employee Salary Trend')
plt.xlabel('Employee Number')
plt.ylabel('Salary')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
